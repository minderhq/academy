```python
import torch
import torch.nn as nn
import matplotlib.pyplot as plt

print("Ready for Deep Learning")
```

## 1. Activation Functions

```python
# TODO: Compare activation functions
x = torch.linspace(-5, 5, 100)

activations = {
    "ReLU": nn.ReLU()(x),
    "Sigmoid": torch.sigmoid(x),
    "Tanh": torch.tanh(x),
    "GELU": nn.GELU()(x),
}

# Plot
fig, axes = plt.subplots(2, 2, figsize=(10, 8))
axes = axes.flatten()

for ax, (name, y) in zip(axes, activations.items()):
    ax.plot(x.numpy(), y.numpy())
    ax.set_title(name)
    ax.grid(True)

plt.tight_layout()
plt.show()
```

## 2. Loss Functions

```python
# TODO: Compare loss functions
predictions = torch.tensor([0.2, 0.7, 0.9, 0.4])
targets = torch.tensor([0.0, 1.0, 1.0, 0.0])

# MSELoss
mse_loss = nn.MSELoss()
mse = mse_loss(predictions, targets)

# CrossEntropy
ce_loss = nn.BCELoss()
ce = ce_loss(predictions, targets)

print(f"MSE Loss: {mse:.4f}")
print(f"BCE Loss: {ce:.4f}")
```

## 3. Optimizers

```python
# TODO: Compare optimizers
model = nn.Linear(10, 1)

optimizers = {
    "SGD": torch.optim.SGD(model.parameters(), lr=0.01),
    "Adam": torch.optim.Adam(model.parameters(), lr=0.01),
    "AdamW": torch.optim.AdamW(model.parameters(), lr=0.01),
}

print(f"Available optimizers: {list(optimizers.keys())}")
```

## 4. Regularization

```python
# TODO: Apply dropout
class RegularizedNet(nn.Module):
    def __init__(self):
        super().__init__()
        self.fc1 = nn.Linear(784, 256)
        self.dropout = nn.Dropout(0.5)
        self.fc2 = nn.Linear(256, 10)
        self.relu = nn.ReLU()

    def forward(self, x):
        x = x.view(x.size(0), -1)
        x = self.relu(self.fc1(x))
        x = self.dropout(x)
        x = self.fc2(x)
        return x

model = RegularizedNet()
print(model)
```
