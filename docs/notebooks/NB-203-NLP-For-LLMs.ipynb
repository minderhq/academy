```python
from transformers import AutoTokenizer
import torch

print("Libraries imported")
```

## 1. Tokenization

```python
# TODO: Load tokenizer
tokenizer = AutoTokenizer.from_pretrained("gpt2")

text = "Hello, world! This is a test."

# Tokenize
tokens = tokenizer.tokenize(text)
token_ids = tokenizer.convert_tokens_to_ids(tokens)

print(f"Text: {text}")
print(f"Tokens: {tokens}")
print(f"Token IDs: {token_ids}")
print(f"Vocab size: {len(tokenizer)}")
```

## 2. Special Tokens

```python
# TODO: Special tokens
print(f"PAD token: {tokenizer.pad_token}")
print(f"EOS token: {tokenizer.eos_token}")
print(f"BOS token: {tokenizer.bos_token}")
print(f"UNK token: {tokenizer.unk_token}")

# Encode with special tokens
encoded = tokenizer(text, return_tensors="pt")
print(f"\nEncoded shape: {encoded['input_ids'].shape}")
print(f"Attention mask: {encoded['attention_mask']}")
```

## 3. Text Preprocessing

```python
import re

def preprocess_text(text):
    """Basic text preprocessing."""

    # TODO: Clean text
    # Lowercase
    text = text.lower()

    # Remove extra whitespace
    text = " ".join(text.split())

    # Remove special characters (keep punctuation)
    text = re.sub(r"[^a-zA-Z0-9\s\.,!?'"]", "", text)

    return text

# Test
messy = "  Hello!!!   This is a TEST...   "
clean = preprocess_text(messy)
print(f"Original: '{messy}'")
print(f"Cleaned: '{clean}'")
```

## 4. Perplexity Calculation

```python
def calculate_perplexity(model, tokenizer, text):
    """Calculate perplexity of text."""

    # TODO: Encode text
    inputs = tokenizer(text, return_tensors="pt")

    # Get logits
    with torch.no_grad():
        outputs = model(**inputs, labels=inputs["input_ids"])

    # Calculate perplexity
    loss = outputs.loss.item()
    perplexity = torch.exp(torch.tensor(loss))

    return perplexity.item()

# Test
# perplexity = calculate_perplexity(model, tokenizer, "The quick brown fox")
# print(f"Perplexity: {perplexity:.2f}")

print("Perplexity function defined")
```
